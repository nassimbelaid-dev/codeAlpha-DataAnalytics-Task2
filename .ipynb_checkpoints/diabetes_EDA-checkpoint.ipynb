# Exploratory Data Analysis (EDA) — Pima Indians Diabetes Dataset

Ce notebook a pour unique objectif de réaliser une **analyse exploratoire des données (EDA)**
sur le dataset `diabetes.csv`.

Aucun modèle de Machine Learning n'est entraîné dans ce notebook : l'objectif est uniquement
de comprendre la structure des données, d'identifier des tendances, des anomalies et des
problèmes de qualité, et de formuler / tester des hypothèses simples à l'aide de statistiques
et de visualisations.


## 1. Import Libraries

On importe uniquement les librairies nécessaires à l'analyse exploratoire (pandas, numpy, matplotlib, seaborn). Aucune librairie de Machine Learning (scikit-learn) n'est nécessaire ici.

```
# Librairies pour la manipulation de données
import pandas as pd
import numpy as np

# Librairies pour la visualisation
import matplotlib.pyplot as plt
import seaborn as sns
from pandas.plotting import scatter_matrix

# Réglages d'affichage : on améliore la lisibilité des graphiques
sns.set(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)
```

## 2. Load Dataset

On charge le dataset `diabetes.csv`. Le fichier est supposé se trouver dans le même dossier que ce notebook ; adapter le chemin si nécessaire.

```
# Chargement du dataset
# NOTE : adapter le chemin ci-dessous si le fichier diabetes.csv se trouve ailleurs
df = pd.read_csv('diabetes.csv')

# On affiche les premières lignes pour vérifier que le chargement s'est bien passé
df.head()
```

## 3. Questions before the analysis

Avant de commencer toute analyse, il est important de formuler les questions auxquelles
on souhaite répondre. Cela permet de guider l'exploration et d'éviter une analyse "à l'aveugle".

Questions posées pour ce dataset :

1. Combien d'observations et de variables le dataset contient-il ?
2. Quelle est la variable cible (target) ?
3. Y a-t-il des valeurs manquantes ?
4. Y a-t-il des doublons ?
5. Y a-t-il des valeurs aberrantes (outliers) ?
6. Le dataset est-il équilibré entre les deux classes (diabétique / non-diabétique) ?
7. Quelles variables semblent liées au diabète ?
8. Quelles variables pourraient nécessiter un pré-traitement (normalisation, imputation, etc.) ?

Ces questions seront traitées tout au long du notebook, section par section.


## 4. Data Structure Exploration

On explore la structure générale du dataset : dimensions, colonnes, types de données. Cela répond directement aux questions 1 et 2 posées plus haut.

```
# Dimensions du dataset : nombre de lignes (observations) et de colonnes (variables)
print("Dimensions du dataset (lignes, colonnes) :", df.shape)
```

```
# Liste des colonnes du dataset
print("Colonnes du dataset :")
print(df.columns.tolist())
```

```
# Informations générales : types de données, valeurs non nulles par colonne
df.info()
```

```
# Types de données de chaque variable
# Utile pour vérifier qu'aucune variable numérique n'est stockée comme texte par erreur
df.dtypes
```

**Conclusion (Data Structure) :**
Le dataset contient 768 observations et 9 variables, dont 8 variables explicatives quantitatives
(Pregnancies, Glucose, BloodPressure, SkinThickness, Insulin, BMI, DiabetesPedigreeFunction, Age)
et une variable cible binaire `Outcome` (0 = non diabétique, 1 = diabétique). Toutes les colonnes
sont de type numérique, ce qui facilite l'analyse statistique.


## 5. Descriptive Statistics

On calcule les statistiques descriptives (moyenne, écart-type, médiane, min, max) pour chaque variable quantitative. Cela permet de repérer des échelles très différentes entre variables et de potentielles valeurs aberrantes.

```
# On définit les variables explicatives (toutes les colonnes sauf la cible Outcome)
attributes_Quant = df.columns[:-1]

# Statistiques descriptives détaillées, variable par variable
for colonne in attributes_Quant:
    print(f"--- {colonne} ---")
    print(f" moyenne     : {df[colonne].mean():.2f}")
    print(f" ecart_type  : {df[colonne].std():.2f}")
    print(f" median      : {df[colonne].median():.2f}")
    print(f" min         : {df[colonne].min()}")
    print(f" max         : {df[colonne].max()}")
    print()
```

```
# Résumé statistique global : équivalent condensé de la boucle précédente
df.describe(include='all')
```

**Conclusion (Descriptive Statistics) :**
Les variables ne sont pas sur la même échelle (par exemple Insulin varie de 0 à plusieurs centaines,
alors que DiabetesPedigreeFunction reste entre 0 et 2). Certains écarts-types élevés (Insulin,
SkinThickness) suggèrent la présence de valeurs extrêmes, qu'on explorera avec les boxplots plus loin.


## 6. Class Distribution

On regarde la répartition de la variable cible `Outcome` pour savoir si le dataset est équilibré (question 6).

```
# Nombre d'observations par classe (0 = non diabétique, 1 = diabétique)
frequence = df['Outcome'].value_counts()
print("Fréquence par classe :")
print(frequence)

# Même information en pourcentage, plus facile à interpréter
proportions = df['Outcome'].value_counts(normalize=True) * 100
print("\nProportions (%) :")
print(proportions)
```

```
# Visualisation de la répartition des classes
plt.figure(figsize=(6, 5))
sns.countplot(x='Outcome', data=df)
plt.xlabel('Outcome (0 = Non-diabétique, 1 = Diabétique)')
plt.ylabel('Nombre de patients')
plt.title('Répartition des classes (Outcome)')
plt.show()
```

**Conclusion (Class Distribution) :**
Le dataset est déséquilibré : environ 65% des patients sont non-diabétiques (Outcome = 0)
contre 35% de diabétiques (Outcome = 1). Ce déséquilibre est important à garder en tête pour une
éventuelle phase de modélisation future (ce n'est pas traité ici, l'objectif étant uniquement l'EDA).


## 7. Correlation Analysis

On étudie les relations entre variables à l'aide des matrices de covariance et de corrélation, puis on les visualise avec une heatmap. Cela permet de repérer les variables les plus liées entre elles, et notamment celles liées à `Outcome` (question 7).

```
# Matrice de covariance : montre comment les variables varient ensemble (dépend de l'échelle)
covariance = df[attributes_Quant].cov()
print("Matrice de covariance :")
covariance
```

```
# Matrice de corrélation : version normalisée (entre -1 et 1), plus facile à interpréter
correlation = df[attributes_Quant].corr()
print("Matrice de corrélation :")
correlation
```

```
# Visualisation de la matrice de corrélation sous forme de heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(correlation, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Matrice de corrélation')
plt.show()
```

**Conclusion (Correlation Analysis) :**
Aucune paire de variables explicatives n'est fortement corrélée entre elles (pas de redondance
majeure), ce qui est plutôt positif. Les variables qui semblent le plus liées entre elles sont
Age/Pregnancies et SkinThickness/Insulin, ce qui est cohérent d'un point de vue physiologique.


## 8. Data Visualization

On visualise la distribution des variables, les relations entre elles et les éventuelles valeurs aberrantes, afin de répondre aux questions 3, 5 et 7.

```
# Distribution de la variable Glucose
plt.figure(figsize=(10, 6))
plt.hist(df['Glucose'], bins=8, edgecolor='black')
plt.xlabel('Glucose')
plt.ylabel('Fréquence')
plt.title('Distribution du Glucose')
plt.show()
```

```
# Distribution de quelques autres variables clés (BMI et Age)
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].hist(df['BMI'], bins=8, edgecolor='black', color='skyblue')
axes[0].set_title('Distribution du BMI')
axes[0].set_xlabel('BMI')
axes[0].set_ylabel('Fréquence')

axes[1].hist(df['Age'], bins=8, edgecolor='black', color='salmon')
axes[1].set_title("Distribution de l'Age")
axes[1].set_xlabel('Age')
axes[1].set_ylabel('Fréquence')

plt.tight_layout()
plt.show()
```

```
# Matrice de nuages de points : permet de visualiser les relations deux à deux entre variables
fig = scatter_matrix(df[attributes_Quant], figsize=(15, 15), diagonal='hist')
plt.suptitle('Matrice de nuages de points', y=1.02)
plt.show()
```

```
# Boxplot global des variables explicatives : utile pour repérer visuellement les valeurs
# aberrantes (points en dehors des moustaches) et les différences d'échelle entre variables
X = df.drop(columns=['Outcome'])

fig, ax = plt.subplots(figsize=(12, 6))
ax.boxplot(X.values, labels=X.columns)
ax.set_title("Boxplot des variables explicatives")
ax.set_ylabel("Valeurs")
plt.xticks(rotation=45)
plt.show()
```

**Conclusion (Data Visualization) :**
Glucose et BMI semblent suivre une distribution proche de la normale (avec un léger étalement
vers la droite), tandis qu'Insulin et SkinThickness montrent de nombreux points au-delà des
moustaches du boxplot, signe probable de valeurs aberrantes ou de valeurs manquantes codées par 0
(voir section Data Quality Assessment).


## 9. Hypothesis Testing

On formule quelques hypothèses simples et intuitives, puis on les vérifie à l'aide de statistiques descriptives et de visualisations (groupby, moyennes, boxplots).

### Hypothesis 1

**People with diabetes tend to have higher glucose levels.**


```
# On compare la moyenne de Glucose entre les deux classes (diabétique / non-diabétique)
moyennes_glucose = df.groupby('Outcome')['Glucose'].mean()
print("Glucose moyen par classe :")
print(f" Non-diabétique (0) : {moyennes_glucose[0]:.2f} mg/dL")
print(f" Diabétique (1)     : {moyennes_glucose[1]:.2f} mg/dL")
```

```
# Visualisation par boxplot pour confirmer visuellement la différence entre les deux groupes
plt.figure(figsize=(6, 5))
sns.boxplot(x='Outcome', y='Glucose', data=df)
plt.xlabel('Outcome (0 = Non-diabétique, 1 = Diabétique)')
plt.ylabel('Glucose')
plt.title('Glucose selon le statut diabétique')
plt.show()
```

**Conclusion Hypothèse 1 :** L'hypothèse est confirmée : le taux moyen de glucose est nettement plus élevé chez les patients diabétiques, ce qui est cohérent avec la physiologie du diabète.

### Hypothesis 2

**Diabetic patients tend to have a higher BMI.**


```
# Comparaison du BMI moyen entre les deux classes
moyennes_bmi = df.groupby('Outcome')['BMI'].mean()
print("BMI moyen par classe :")
print(f" Non-diabétique (0) : {moyennes_bmi[0]:.2f}")
print(f" Diabétique (1)     : {moyennes_bmi[1]:.2f}")
```

```
# Visualisation par boxplot
plt.figure(figsize=(6, 5))
sns.boxplot(x='Outcome', y='BMI', data=df)
plt.xlabel('Outcome (0 = Non-diabétique, 1 = Diabétique)')
plt.ylabel('BMI')
plt.title('BMI selon le statut diabétique')
plt.show()
```

**Conclusion Hypothèse 2 :** L'hypothèse est confirmée : les patients diabétiques ont en moyenne un BMI plus élevé que les non-diabétiques, suggérant un lien entre surpoids et diabète.

### Hypothesis 3

**Older people are more likely to have diabetes.**


```
# Comparaison de l'âge moyen entre les deux classes
moyennes_age = df.groupby('Outcome')['Age'].mean()
print("Age moyen par classe :")
print(f" Non-diabétique (0) : {moyennes_age[0]:.2f} ans")
print(f" Diabétique (1)     : {moyennes_age[1]:.2f} ans")
```

```
# Visualisation par boxplot
plt.figure(figsize=(6, 5))
sns.boxplot(x='Outcome', y='Age', data=df)
plt.xlabel('Outcome (0 = Non-diabétique, 1 = Diabétique)')
plt.ylabel('Age')
plt.title('Age selon le statut diabétique')
plt.show()
```

**Conclusion Hypothèse 3 :** L'hypothèse est globalement confirmée : les patients diabétiques sont en moyenne plus âgés, même si l'écart est moins marqué que pour Glucose ou BMI.

## 10. Data Quality Assessment

On recherche ici les problèmes potentiels dans les données : valeurs manquantes, doublons, et valeurs à 0 suspectes (question 3, 4, 5, 8).

```
# Recherche de valeurs manquantes (NaN) pour chaque colonne
print("Nombre de valeurs manquantes (NaN) par colonne :")
df.isnull().sum()
```

```
# Recherche de lignes dupliquées dans le dataset
print("Nombre de lignes dupliquées :", df.duplicated().sum())
```

```
# Certaines colonnes ne peuvent physiologiquement pas valoir 0 (Glucose, BloodPressure, BMI,
# Insulin, SkinThickness). Une valeur de 0 dans ces colonnes est donc probablement une valeur
# manquante codée par 0, plutôt qu'une vraie mesure.
colonnes_suspectes = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']

print("Nombre de valeurs égales à 0 (potentiellement manquantes) par colonne :")
for col in colonnes_suspectes:
    nb_zeros = (df[col] == 0).sum()
    pourcentage = nb_zeros / len(df) * 100
    print(f" {col:15s} : {nb_zeros} valeurs à 0 ({pourcentage:.1f}%)")
```

**Conclusion (Data Quality Assessment) :**
Le dataset ne contient ni valeurs manquantes explicites (NaN) ni doublons. En revanche, plusieurs
colonnes contiennent des valeurs égales à 0 qui n'ont pas de sens physiologique (par exemple
Insulin et SkinThickness comptent un nombre important de zéros). Ces zéros représentent très
probablement des valeurs manquantes codées différemment, et devront être traités (par exemple
par imputation) avant toute étape de modélisation future.


## 11. Final Conclusions

### Final EDA Summary

**Tendances observées :**
- Le glucose, le BMI et l'âge sont plus élevés en moyenne chez les patients diabétiques.
- Les distributions de Glucose et BMI sont proches d'une distribution normale, légèrement étalées à droite.

**Corrélations importantes :**
- Aucune corrélation très forte entre variables explicatives (pas de redondance majeure).
- Des corrélations modérées existent entre Age/Pregnancies et entre SkinThickness/Insulin.

**Anomalies détectées :**
- Présence de valeurs extrêmes (outliers) sur Insulin et SkinThickness, visibles sur les boxplots.
- Écarts-types élevés sur certaines variables, révélateurs de valeurs très dispersées.

**Problèmes de qualité des données :**
- Pas de valeurs manquantes (NaN) ni de doublons.
- Présence de valeurs à 0 suspectes sur Glucose, BloodPressure, SkinThickness, Insulin et BMI,
  qui représentent probablement des valeurs manquantes codées par zéro.

**Déséquilibre des classes :**
- Le dataset est déséquilibré (~65% non-diabétiques vs ~35% diabétiques).

**Points intéressants pour une éventuelle phase de Machine Learning (non traités ici) :**
- Traiter les zéros suspects (imputation) avant tout entraînement de modèle.
- Envisager une normalisation des variables, celles-ci n'étant pas sur la même échelle.
- Tenir compte du déséquilibre des classes (par exemple via un ré-échantillonnage ou une
  pondération des classes) lors d'une future étape de modélisation.
- Glucose, BMI et Age apparaissent comme des variables potentiellement discriminantes pour
  prédire le diabète.
